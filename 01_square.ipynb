## 1. 첫 애니메이션 — 중심과 원

Manim 코드는 항상 이런 구조예요.

```
%%manim -qm -v WARNING 장면이름     ← Colab에게 "애니메이션 만들어줘"라고 알리는 명령
class 장면이름(Scene):              ← 하나의 애니메이션(장면)
    def construct(self):           ← 장면 안에서 일어날 일을 적는 곳
        ...                        ← 도형 만들기 + 움직이기
```

아래 셀을 실행하면 **중심 O**가 찍히고, 그 둘레에 **원**이 그려져요.

````python
%%manim -qm -v WARNING CenterAndCircle

class CenterAndCircle(Scene):
    def construct(self):
        center = Dot(ORIGIN, color=ORANGE)                 # 원점에 주황색 점(중심)
        label_O = Text("O", font_size=32).next_to(center, DOWN + LEFT, buff=0.1)
        circle = Circle(radius=2.5, color=WHITE)           # 반지름 2.5인 하얀 원

        self.play(FadeIn(center), Write(label_O))          # 중심이 나타남
        self.wait(0.5)
        self.play(Create(circle), run_time=2)              # 원이 그려짐
        self.wait(1)
````

## 2. 도형 만들기 — 점, 선분, 원, 글자

| 코드 | 도형 |
|---|---|
| `Dot([x, y, 0])` | 점 |
| `Line(시작점, 끝점)` | 선분 |
| `Circle(radius=2)` | 원 |
| `Text("내용", font="NanumGothic")` | 글자 (한글은 꼭 글꼴 지정!) |

- 위치 옮기기: `.shift(UP)`, `.shift(LEFT * 3)` — `UP, DOWN, LEFT, RIGHT`
- 색: `RED, ORANGE, YELLOW, GREEN, BLUE, PURPLE, PINK, WHITE`
- 채우기: `.set_fill(색, opacity=0.5)` — opacity는 0(투명)~1(불투명)

아래 셀은 중심 O에서 원 위의 한 점 P까지 **반지름**을 긋고, 이름을 붙여요.

````python
%%manim -qm -v WARNING RadiusAndName

class RadiusAndName(Scene):
    def construct(self):
        title = Text("원의 중심과 반지름", font="NanumGothic", font_size=40).to_edge(UP)
        self.play(Write(title))

        O = ORIGIN
        P = [2.5, 0, 0]                                    # 원 위의 한 점 (오른쪽)
        circle = Circle(radius=2.5, color=WHITE)
        center = Dot(O, color=ORANGE)
        point  = Dot(P, color=YELLOW)
        radius = Line(O, P, color=YELLOW)                  # 반지름 OP

        t_O = Text("O", font_size=30).next_to(center, DOWN + LEFT, buff=0.1)
        t_P = Text("P", font_size=30, color=YELLOW).next_to(point, RIGHT, buff=0.1)
        t_r = Text("r", font_size=30, color=YELLOW).next_to(radius, UP, buff=0.1)

        self.play(Create(circle), FadeIn(center), Write(t_O))
        self.play(FadeIn(point), Write(t_P))
        self.play(Create(radius), Write(t_r))
        self.wait(0.5)

        # 원 안을 살짝 채워 보기
        self.play(circle.animate.set_fill(BLUE, opacity=0.3))
        self.wait(1)
````

## 3. ⭐ 말없는 증명 — 같은 거리의 점들이 모이면?

시작의 광장 게임에서 본 것처럼, **중심에서 같은 거리(r)에 있는 점**을 하나씩 찍어 볼게요.

- 각도 `deg`를 0°, 30°, 60°, … 로 바꾸면서 점의 위치를 계산: `(r·cos θ, r·sin θ)`
- 점이 12개, 36개, 360개… 점점 많아지면 → **원**이 나타나요!

> 이것이 원의 정의예요. "원 = 한 점에서 같은 거리에 있는 **모든** 점의 모임"

````python
%%manim -qm -v WARNING PointsBecomeCircle

class PointsBecomeCircle(Scene):
    def construct(self):
        r = 2.5
        O = ORIGIN
        title = Text("중심에서 같은 거리의 점들을 모으면?", font="NanumGothic", font_size=34).to_edge(UP)
        center = Dot(O, color=ORANGE)
        t_O = Text("O", font_size=30).next_to(center, DOWN + LEFT, buff=0.1)
        self.play(Write(title), FadeIn(center), Write(t_O))

        # 1단계: 점 12개 — 자(반지름)를 돌리며 하나씩 찍기
        ruler = Line(O, [r, 0, 0], color=YELLOW)           # 길이 r인 자
        self.play(Create(ruler))
        dots = VGroup()
        for deg in range(0, 360, 30):                      # 0, 30, 60, ..., 330
            p = [r * np.cos(deg * DEGREES), r * np.sin(deg * DEGREES), 0]
            new_ruler = Line(O, p, color=YELLOW)
            self.play(Transform(ruler, new_ruler), run_time=0.25)
            d = Dot(p, color=YELLOW)
            dots.add(d)
            self.play(FadeIn(d), run_time=0.15)
        self.play(FadeOut(ruler))
        self.wait(0.5)

        # 2단계: 점을 36개, 360개로 늘리기
        for n, size in [(36, 0.07), (360, 0.04)]:
            more = VGroup(*[Dot([r * np.cos(k * TAU / n), r * np.sin(k * TAU / n), 0],
                                radius=size, color=YELLOW) for k in range(n)])
            msg = Text(f"점 {n}개", font="NanumGothic", font_size=30, color=YELLOW).to_edge(DOWN)
            self.play(FadeIn(more), FadeIn(msg), run_time=1)
            self.wait(0.8)
            self.play(FadeOut(msg), run_time=0.3)

        # 3단계: 진짜 원으로!
        circle = Circle(radius=r, color=WHITE, stroke_width=6)
        result = Text("원 = 한 점에서 같은 거리에 있는 점들의 모임",
                      font="NanumGothic", font_size=30, color=WHITE).to_edge(DOWN)
        self.play(Create(circle), Write(result), run_time=2)
        self.wait(2)
````

## 4. 반지름이 다르면? — 원의 크기

같은 중심에서 거리만 다르면 **크기가 다른 원**이 돼요. 반복문으로 반지름 1, 2, 3인 원을 그려 봅시다.

> 💡 `for r in [1, 2, 3]:` — 대괄호 안의 수를 차례로 r에 넣어 반복해요.

````python
%%manim -qm -v WARNING ThreeCircles

class ThreeCircles(Scene):
    def construct(self):
        center = Dot(ORIGIN, color=ORANGE)
        self.play(FadeIn(center))

        colors = [YELLOW, GREEN, BLUE]
        for r, color in zip([1, 2, 3], colors):            # r=1 노랑, r=2 초록, r=3 파랑
            circle = Circle(radius=r, color=color)
            label = Text(f"r = {r}", font_size=28, color=color).next_to(circle, RIGHT, buff=0.2)
            self.play(Create(circle), Write(label))
        self.wait(1.5)
````

## 🏆 도전 과제 — 나만의 '시작의 광장' 만들기

아래 셀을 고쳐서 **자기 이름이 들어간 원의 정의 영상**을 완성해 보세요.

1. 제목의 `홍길동`을 자기 이름으로 바꾸기
2. 반지름 `r`을 다른 값으로 바꾸고, 점 개수 `n`을 8 → 24로 늘려 보기
3. 마지막에 원 안을 좋아하는 색으로 채우기 (`set_fill`)
4. (보너스) 중심 O를 `LEFT * 2`로 옮겨서 원 전체가 왼쪽으로 가게 하기

````python
%%manim -qm -v WARNING MySquareOfBeginning

class MySquareOfBeginning(Scene):
    def construct(self):
        # 1. 이름을 자기 이름으로 바꾸세요
        title = Text("홍길동의 시작의 광장", font="NanumGothic", font_size=40).to_edge(UP)
        self.play(Write(title))

        # 2. 반지름과 점 개수를 바꿔 보세요
        r = 2
        n = 8
        O = ORIGIN                       # 4. (보너스) LEFT * 2 로 바꿔 보세요
        center = Dot(O, color=ORANGE)
        self.play(FadeIn(center))

        dots = VGroup(*[Dot(O + np.array([r * np.cos(k * TAU / n), r * np.sin(k * TAU / n), 0]),
                            color=YELLOW) for k in range(n)])
        self.play(LaggedStart(*[FadeIn(d) for d in dots], lag_ratio=0.2))

        circle = Circle(radius=r, color=WHITE).move_to(O)
        self.play(Create(circle))

        # 3. 원 안을 채워 보세요: circle.animate.set_fill(색, opacity=0.4)

        self.wait(2)
````

## 📝 오늘 배운 것 정리

- `Scene` 클래스 안의 `construct`에 장면을 적는다.
- `Dot, Line, Circle, Text`로 도형과 글자를 만들고, `self.play(Create/FadeIn/Write)`로 움직인다.
- 원 위의 점은 `(r·cos θ, r·sin θ)`로 구한다 → 각도를 바꾸며 반복하면 점이 원을 만든다.
- **원의 정의**: 한 점(중심)에서 같은 거리(반지름)에 있는 모든 점의 모임.
- 한글은 `font="NanumGothic"`!
