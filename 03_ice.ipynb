## 1. 수직이등분선 작도 — 컴퍼스와 자를 코드로

선분 AB의 **수직이등분선**은 이렇게 작도해요.

1. A를 중심으로 반지름이 충분히 큰 원을 그린다.
2. B를 중심으로 **같은 반지름**의 원을 그린다.
3. 두 원이 만나는 두 점을 직선으로 잇는다.

코드에서 컴퍼스는 `Circle`, 자는 `Line`이에요. `DashedVMobject`는 점선으로 바꿔 줘요.

```python
%%manim -qm -v WARNING PerpBisector

class PerpBisector(Scene):
    def construct(self):
        A, B = np.array([-2, 0, 0]), np.array([2, 0, 0])
        r = 3
        seg = Line(A, B)
        self.play(Create(seg), FadeIn(Dot(A)), FadeIn(Dot(B)),
                  Write(Text("A", font_size=28).next_to(A, DOWN + LEFT)),
                  Write(Text("B", font_size=28).next_to(B, DOWN + RIGHT)))

        c1 = DashedVMobject(Circle(radius=r, color=BLUE).move_to(A))
        c2 = DashedVMobject(Circle(radius=r, color=GREEN).move_to(B))
        self.play(Create(c1)); self.play(Create(c2))

        h = np.sqrt(r ** 2 - 2 ** 2)              # 교점의 높이
        P, Q = np.array([0, h, 0]), np.array([0, -h, 0])
        self.play(FadeIn(Dot(P, color=YELLOW)), FadeIn(Dot(Q, color=YELLOW)))
        bis = Line(P * 1.4, Q * 1.4, color=YELLOW)
        self.play(Create(bis))

        M = Dot(ORIGIN, color=ORANGE)
        ra = RightAngle(bis, seg, length=0.3, color=ORANGE)
        msg = Text("AM = MB, 그리고 수직!", font="NanumGothic", font_size=30, color=YELLOW).to_edge(UP)
        self.play(FadeIn(M), Create(ra), Write(msg))
        self.wait(2)
```

## 2. ⭐ 말없는 증명 — 세 수직이등분선은 한 점에서 만난다

삼각형 ABC의 세 변에 각각 수직이등분선을 그으면 **한 점 O**에서 만나요. 이 점이 **외심**이에요.

왜 그럴까요? 수직이등분선 위의 점은 양 끝점에서 같은 거리에 있어요.
→ AB의 수직이등분선 위에 있으면 OA = OB, BC의 수직이등분선 위에 있으면 OB = OC.
→ 따라서 **OA = OB = OC**: O를 중심으로 세 꼭짓점을 지나는 원(외접원)을 그릴 수 있어요!

> 💡 코드에서 외심은 수식 대신 `line_intersection`으로 두 수직이등분선의 교점을 구했어요.

```python
%%manim -qm -v WARNING Circumcenter

def perp_bisector(P, Q, length=9):
    mid = (P + Q) / 2
    d = Q - P
    n = np.array([-d[1], d[0], 0]) / np.linalg.norm(d)     # 수직 방향
    return Line(mid - n * length / 2, mid + n * length / 2)

class Circumcenter(Scene):
    def construct(self):
        A, B, C = np.array([-3, -1.5, 0]), np.array([2.5, -1.5, 0]), np.array([0.5, 2.2, 0])
        tri = Polygon(A, B, C, color=WHITE)
        names = VGroup(Text("A", font_size=28).next_to(A, DL, buff=0.1),
                       Text("B", font_size=28).next_to(B, DR, buff=0.1),
                       Text("C", font_size=28).next_to(C, UP, buff=0.1))
        self.play(Create(tri), Write(names))

        colors = [BLUE, GREEN, PINK]
        lines = []
        for (P, Q), color in zip([(A, B), (B, C), (C, A)], colors):
            mid = Dot((P + Q) / 2, color=color)
            l = perp_bisector(P, Q).set_color(color)
            self.play(FadeIn(mid), Create(DashedVMobject(l)), run_time=1)
            lines.append(l)

        O = line_intersection(lines[0].get_start_and_end(), lines[1].get_start_and_end())
        dotO = Dot(O, color=ORANGE).scale(1.3)
        tO = Text("외심 O", font="NanumGothic", font_size=28, color=ORANGE).next_to(dotO, RIGHT, buff=0.1)
        self.play(FadeIn(dotO), Write(tO))
        self.wait(0.5)

        # OA = OB = OC 를 반지름으로 보여 주고 외접원 그리기
        R = np.linalg.norm(A - O)
        radii = VGroup(*[Line(O, P, color=ORANGE) for P in [A, B, C]])
        self.play(Create(radii))
        circ = Circle(radius=R, color=YELLOW).move_to(O)
        msg = Text("OA = OB = OC  → 외접원", font="NanumGothic", font_size=32, color=YELLOW).to_edge(UP)
        self.play(Create(circ), Write(msg), run_time=2)
        self.wait(2)
```

## 3. 직각삼각형의 외심은 빗변의 중점

∠C = 90°인 직각삼각형에서는 외심이 **빗변 AB의 한가운데**에 와요.
그래서 **빗변이 곧 외접원의 지름**이 됩니다. (얼음의 도시 영상 '삼각형과 외접원의 지름'이 바로 이 내용!)

아래 코드는 점 C를 원 위에서 움직여도 ∠C가 항상 90°임을 보여 줘요.

```python
%%manim -qm -v WARNING RightTriangleCircum

class RightTriangleCircum(Scene):
    def construct(self):
        R = 2.5
        A, B = np.array([-R, 0, 0]), np.array([R, 0, 0])      # 지름 AB
        circ = Circle(radius=R, color=GREY)
        O = Dot(ORIGIN, color=ORANGE)
        diam = Line(A, B, color=YELLOW)
        self.play(Create(circ), Create(diam), FadeIn(O),
                  Write(Text("A", font_size=28).next_to(A, LEFT)), Write(Text("B", font_size=28).next_to(B, RIGHT)),
                  Write(Text("O", font_size=28).next_to(O, DOWN)))

        t = ValueTracker(60)          # C의 각도 위치
        def C_pos():
            a = t.get_value() * DEGREES
            return R * np.array([np.cos(a), np.sin(a), 0])
        tri = always_redraw(lambda: Polygon(A, B, C_pos(), color=BLUE, fill_opacity=0.25))
        ra = always_redraw(lambda: RightAngle(Line(C_pos(), A), Line(C_pos(), B), length=0.35, color=RED))
        dotC = always_redraw(lambda: Dot(C_pos(), color=BLUE))
        label = always_redraw(lambda: Text("∠C = 90°", font_size=30, color=RED).next_to(dotC, UP))
        self.add(tri, ra, dotC, label)
        self.play(t.animate.set_value(150), run_time=3)
        self.play(t.animate.set_value(30), run_time=3)
        msg = Text("빗변 AB가 지름 → 외심은 빗변의 중점", font="NanumGothic", font_size=30, color=YELLOW).to_edge(DOWN)
        self.play(Write(msg))
        self.wait(2)
```

## 🏆 도전 과제 — 외심은 어디로 갈까?

`Circumcenter` 코드를 복사한 아래 셀에서 꼭짓점 **C의 좌표**를 바꿔 보세요.

1. `C = [0.5, 2.2, 0]` → `[0.5, 0.6, 0]` (둔각삼각형): 외심이 삼각형 **밖**으로 나가나요?
2. C를 `[-3, 2, 0]`으로 바꾸면 (∠A = 90°) 외심이 어디에 있나요?
3. (보너스) 세 변의 수직이등분선 색을 바꾸고, 제목에 자기 이름을 넣어 보세요.

```python
%%manim -qm -v WARNING MyCircumcenter

def perp_bisector(P, Q, length=12):
    mid = (P + Q) / 2
    d = Q - P
    n = np.array([-d[1], d[0], 0]) / np.linalg.norm(d)
    return Line(mid - n * length / 2, mid + n * length / 2)

class MyCircumcenter(Scene):
    def construct(self):
        A = np.array([-3, -1.5, 0])
        B = np.array([2.5, -1.5, 0])
        C = np.array([0.5, 2.2, 0])          # ← 여기를 바꿔 보세요
        title = Text("홍길동의 외심 탐구", font="NanumGothic", font_size=34).to_edge(UP)
        tri = Polygon(A, B, C, color=WHITE)
        self.play(Write(title), Create(tri))

        lines = [perp_bisector(P, Q).set_color(color)
                 for (P, Q), color in zip([(A, B), (B, C), (C, A)], [BLUE, GREEN, PINK])]
        self.play(*[Create(DashedVMobject(l)) for l in lines])

        O = line_intersection(lines[0].get_start_and_end(), lines[1].get_start_and_end())
        R = np.linalg.norm(A - O)
        self.play(FadeIn(Dot(O, color=ORANGE).scale(1.3)))
        self.play(Create(Circle(radius=R, color=YELLOW).move_to(O)))
        self.wait(2)
```

## 📝 오늘 배운 것 정리

- 수직이등분선 작도: 두 원(`Circle`)의 교점을 잇는다(`Line`).
- 삼각형 세 변의 수직이등분선은 한 점 **외심 O**에서 만나고, OA = OB = OC → **외접원**.
- 예각삼각형은 외심이 안, 둔각삼각형은 밖, 직각삼각형은 **빗변의 중점**(빗변 = 지름).
- `line_intersection`으로 교점, `RightAngle`로 직각 표시.
