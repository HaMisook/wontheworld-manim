## 1. 접선은 반지름에 수직

원과 **한 점에서만** 만나는 직선을 **접선**, 만나는 점을 **접점**이라고 해요.
접점 T를 지나는 반지름 OT와 접선은 항상 **수직**이에요.

> 💡 접선의 방향은 반지름 방향을 90° 돌린 벡터 `(-y, x)`로 구해요.

```python
%%manim -qm -v WARNING TangentPerp

class TangentPerp(Scene):
    def construct(self):
        r = 2.2
        circ = Circle(radius=r, color=GREY)
        O = ORIGIN
        self.play(Create(circ), FadeIn(Dot(O, color=ORANGE)), Write(Text("O", font_size=28).next_to(O, DOWN)))

        a = 40 * DEGREES
        T = r * np.array([np.cos(a), np.sin(a), 0])
        d = np.array([-np.sin(a), np.cos(a), 0])           # 접선 방향 (반지름에 수직)
        radius = Line(O, T, color=YELLOW)
        tangent = Line(T - 3.5 * d, T + 3.5 * d, color=BLUE)
        self.play(Create(radius), FadeIn(Dot(T, color=YELLOW)), Write(Text("T", font_size=28, color=YELLOW).next_to(T, UR)))
        self.play(Create(tangent))
        ra = RightAngle(Line(T, O), Line(T, T + d), length=0.3, color=RED)
        self.play(Create(ra))
        msg = Text("접선 ⟂ 반지름 OT", font="NanumGothic", font_size=32, color=BLUE).to_edge(UP)
        self.play(Write(msg))
        self.wait(2)
```

## 2. ⭐ 말없는 증명 — 두 접선의 길이는 같다

원 밖의 점 P에서 원에 접선을 그으면 **두 개** 그을 수 있어요. 접점을 A, B라 하면 **PA = PB**.

왜? △PAO와 △PBO는 OA = OB(반지름), OP 공통, ∠A = ∠B = 90° 이므로 합동!
아래 코드는 두 삼각형을 색칠하고, 직선 OP를 축으로 **접어서** 겹치는 모습을 보여 줘요.

```python
%%manim -qm -v WARNING TwoTangents

def tangent_points(P, r):
    # 원점 중심 반지름 r인 원에 대해 점 P에서 그은 접선의 접점 두 개
    d = np.linalg.norm(P[:2]); ang = np.arctan2(P[1], P[0]); da = np.arccos(r / d)
    return [r * np.array([np.cos(ang + s * da), np.sin(ang + s * da), 0]) for s in (1, -1)]

class TwoTangents(Scene):
    def construct(self):
        r = 2
        O = ORIGIN
        P = np.array([4.5, 0, 0])
        circ = Circle(radius=r, color=GREY).shift(LEFT * 1.5)
        O = O + LEFT * 1.5; P = P + LEFT * 1.5
        self.play(Create(circ), FadeIn(Dot(O, color=ORANGE)), FadeIn(Dot(P, color=WHITE)),
                  Write(Text("O", font_size=28).next_to(O, DOWN)), Write(Text("P", font_size=28).next_to(P, RIGHT)))

        A, B = [q + LEFT * 1.5 for q in tangent_points(P - LEFT * 1.5, r)]
        ta, tb = Line(P, A, color=BLUE, stroke_width=6), Line(P, B, color=GREEN, stroke_width=6)
        self.play(Create(ta), Create(tb), FadeIn(Dot(A, color=BLUE)), FadeIn(Dot(B, color=GREEN)),
                  Write(Text("A", font_size=28, color=BLUE).next_to(A, UP)), Write(Text("B", font_size=28, color=GREEN).next_to(B, DOWN)))

        tri1 = Polygon(P, A, O, color=BLUE, fill_opacity=0.3)
        tri2 = Polygon(P, B, O, color=GREEN, fill_opacity=0.3)
        self.play(FadeIn(tri1), FadeIn(tri2), Create(Line(O, A, color=BLUE)), Create(Line(O, B, color=GREEN)),
                  Create(RightAngle(Line(A, O), Line(A, P), length=0.25, color=RED)),
                  Create(RightAngle(Line(B, P), Line(B, O), length=0.25, color=RED)))
        self.wait(0.5)
        # OP를 축으로 접기 (위 삼각형을 아래로 뒤집기)
        folded = tri1.copy().set_color(YELLOW)
        self.play(Rotate(folded, angle=PI, axis=RIGHT, about_point=O), run_time=2)
        la, lb = np.linalg.norm(P - A), np.linalg.norm(P - B)
        msg = Text(f"PA = PB = {la:.2f}", font_size=34, color=YELLOW).to_edge(UP)
        self.play(Write(msg))
        self.wait(2)
```

## 3. P가 움직여도 PA = PB

점 P를 원 둘레 바깥으로 움직여 봐요. 접점 A, B는 따라 움직이지만 두 접선의 길이는 **항상** 같아요.
숫자판이 두 길이를 매 순간 보여 줍니다.

```python
%%manim -qm -v WARNING MovingTangents

def tangent_points(P, r):
    d = np.linalg.norm(P[:2]); ang = np.arctan2(P[1], P[0]); da = np.arccos(r / d)
    return [r * np.array([np.cos(ang + s * da), np.sin(ang + s * da), 0]) for s in (1, -1)]

class MovingTangents(Scene):
    def construct(self):
        r = 1.8
        circ = Circle(radius=r, color=GREY)
        self.play(Create(circ), FadeIn(Dot(ORIGIN, color=ORANGE)))
        t = ValueTracker(0)
        def P():
            a = t.get_value() * DEGREES; dist = 3.2 + 0.8 * np.sin(3 * a)      # 거리도 살짝 변하게
            return dist * np.array([np.cos(a), np.sin(a), 0])
        def picture():
            p = P(); A, B = tangent_points(p, r)
            return VGroup(Line(p, A, color=BLUE, stroke_width=6), Line(p, B, color=GREEN, stroke_width=6),
                          Dot(p, color=WHITE), Dot(A, color=BLUE), Dot(B, color=GREEN))
        def panel():
            p = P(); A, B = tangent_points(p, r)
            return Text(f"PA = {np.linalg.norm(p-A):.2f}   PB = {np.linalg.norm(p-B):.2f}", font_size=30).to_edge(UP)
        self.add(always_redraw(picture), always_redraw(panel))
        self.play(t.animate.set_value(360), run_time=8, rate_func=linear)
        self.wait(1)
```

## 🏆 도전 과제 — 비눗방울 터뜨리기 🫧

1. `TwoTangents`에서 P의 위치를 `[4.5, 0, 0]` → `[3, 3, 0]`으로 바꿔 보세요. 접점이 어디로 가나요?
2. 원의 색을 바꿔 비눗방울처럼 만들고(`Circle(..., color=PINK).set_fill(PINK, opacity=0.2)`), PA = PB가 확인되면 원이 **사라지는(터지는)** 연출을 `FadeOut`으로 넣어 보세요.
3. (보너스) `MovingTangents`에서 `dist`를 상수 3으로 바꾸면 P가 어떤 길을 따라 움직이나요?

```python
%%manim -qm -v WARNING BubblePop

def tangent_points(P, r):
    d = np.linalg.norm(P[:2]); ang = np.arctan2(P[1], P[0]); da = np.arccos(r / d)
    return [r * np.array([np.cos(ang + s * da), np.sin(ang + s * da), 0]) for s in (1, -1)]

class BubblePop(Scene):
    def construct(self):
        r = 2
        P = np.array([4.5, 0, 0])                       # 1. 위치를 바꿔 보세요
        bubble = Circle(radius=r, color=PINK).set_fill(PINK, opacity=0.2)   # 2. 비눗방울
        self.play(Create(bubble))
        A, B = tangent_points(P, r)
        self.play(Create(Line(P, A, color=BLUE, stroke_width=6)), Create(Line(P, B, color=GREEN, stroke_width=6)), FadeIn(Dot(P)))
        msg = Text(f"PA = PB = {np.linalg.norm(P-A):.2f}  →  팡!", font="NanumGothic", font_size=34, color=YELLOW).to_edge(UP)
        self.play(Write(msg))
        # 2. 여기서 비눗방울을 터뜨려 보세요: self.play(FadeOut(bubble, scale=1.5))
        self.wait(2)
```

## 📝 오늘 배운 것 정리

- 접선은 접점을 지나는 반지름에 **수직**이다.
- 원 밖의 한 점에서 그은 **두 접선의 길이는 같다** (△PAO ≡ △PBO).
- P가 어디로 움직여도 PA = PB는 변하지 않는다.
- `Rotate(axis=RIGHT)`로 접기, `arccos`로 접점 계산, `always_redraw`로 따라 움직이기.
