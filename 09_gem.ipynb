## 1. 내접사각형의 네 각

원 위의 네 점 A, B, C, D로 사각형을 만들면 **원에 내접하는 사각형**이에요.
네 내각을 재 보면 ∠A + ∠C = 180°, ∠B + ∠D = 180°가 나와요.

```python
%%manim -qm -v WARNING CyclicQuad

def angle_deg(V, A, B):
    u, w = A - V, B - V
    return np.degrees(np.arccos(np.clip(np.dot(u, w) / (np.linalg.norm(u) * np.linalg.norm(w)), -1, 1)))

class CyclicQuad(Scene):
    def construct(self):
        r = 2.6
        on = lambda d: r * np.array([np.cos(d * DEGREES), np.sin(d * DEGREES), 0])
        A, B, C, D = on(200), on(300), on(20), on(110)
        circ = Circle(radius=r, color=GREY)
        quad = Polygon(A, B, C, D, color=WHITE)
        self.play(Create(circ), Create(quad),
                  *[Write(Text(n, font_size=28).move_to(p * 1.15)) for n, p in zip("ABCD", (A, B, C, D))])
        pts = [A, B, C, D]
        colors = [PINK, BLUE, PINK, BLUE]
        labels = []
        for i in range(4):
            V, P1, P2 = pts[i], pts[(i - 1) % 4], pts[(i + 1) % 4]
            ang = Angle(Line(V, P2), Line(V, P1), radius=0.5, color=colors[i])
            deg = angle_deg(V, P1, P2)
            lab = Text(f"{deg:.0f}°", font_size=26, color=colors[i]).move_to(V + 0.9 * (ORIGIN - V) / np.linalg.norm(V))
            labels.append(deg)
            self.play(Create(ang), Write(lab), run_time=0.7)
        t1 = Text(f"∠A + ∠C = {labels[0]:.0f}° + {labels[2]:.0f}° = 180°", font_size=30, color=PINK).to_corner(DL)
        t2 = Text(f"∠B + ∠D = {labels[1]:.0f}° + {labels[3]:.0f}° = 180°", font_size=30, color=BLUE).to_corner(DR)
        self.play(Write(t1), Write(t2))
        self.wait(2)
```

## 2. ⭐ 말없는 증명 — 왜 180°일까?

∠A는 호 BCD에 대한 원주각, ∠C는 호 DAB에 대한 원주각이에요.
두 호를 합치면 원 전체(중심각 360°)이므로, 원주각의 합은 360° ÷ 2 = **180°**!

아래 코드는 두 호를 색칠하고 중심각 두 개를 합쳐 360°가 되는 모습을 보여 줘요.

```python
%%manim -qm -v WARNING WhyOneEighty

class WhyOneEighty(Scene):
    def construct(self):
        r = 2.6
        on = lambda d: r * np.array([np.cos(d * DEGREES), np.sin(d * DEGREES), 0])
        A, B, C, D = on(200), on(300), on(20), on(110)
        circ = Circle(radius=r, color=GREY)
        quad = Polygon(A, B, C, D, color=WHITE)
        self.play(Create(circ), Create(quad),
                  *[Write(Text(n, font_size=28).move_to(p * 1.15)) for n, p in zip("ABCD", (A, B, C, D))])

        arc_BCD = Arc(radius=r, start_angle=300 * DEGREES, angle=170 * DEGREES, color=PINK, stroke_width=10)
        arc_DAB = Arc(radius=r, start_angle=110 * DEGREES, angle=190 * DEGREES, color=BLUE, stroke_width=10)
        angA = Angle(Line(A, B), Line(A, D), radius=0.5, color=PINK)
        angC = Angle(Line(C, D), Line(C, B), radius=0.5, color=BLUE)
        self.play(Create(arc_BCD), Create(angA), Write(Text("∠A ↔ 호 BCD", font="NanumGothic", font_size=26, color=PINK).to_corner(UR)))
        self.play(Create(arc_DAB), Create(angC), Write(Text("∠C ↔ 호 DAB", font="NanumGothic", font_size=26, color=BLUE).to_corner(UR).shift(DOWN * 0.5)))

        # 중심각 두 개 → 360°
        O = ORIGIN
        s1 = Sector(radius=1.1, start_angle=300 * DEGREES, angle=170 * DEGREES, color=PINK, fill_opacity=0.5, stroke_width=0)
        s2 = Sector(radius=1.1, start_angle=110 * DEGREES, angle=190 * DEGREES, color=BLUE, fill_opacity=0.5, stroke_width=0)
        self.play(Create(Line(O, B, color=GREY)), Create(Line(O, D, color=GREY)), FadeIn(s1), FadeIn(s2))
        t = Text("중심각 170° + 190° = 360°", font="NanumGothic", font_size=28).to_corner(DL).shift(UP * 0.6)
        t2 = Text("원주각 ∠A + ∠C = 360° ÷ 2 = 180°", font="NanumGothic", font_size=30, color=YELLOW).to_corner(DL)
        self.play(Write(t)); self.play(Write(t2))
        self.wait(2)
```

## 3. 꼭짓점을 움직여도 합은 180° · 외각 = 이웃한 내각의 대각

점 C를 원 위에서 움직이면 ∠C도 ∠A도 변하지만 **합은 늘 180°**예요.
또, 변 BC를 늘여 만든 **외각**(∠DCE)은 이웃한 내각 ∠C의 대각인 **∠A와 크기가 같아요**.

```python
%%manim -qm -v WARNING MovingVertex

def angle_deg(V, A, B):
    u, w = A - V, B - V
    return np.degrees(np.arccos(np.clip(np.dot(u, w) / (np.linalg.norm(u) * np.linalg.norm(w)), -1, 1)))

class MovingVertex(Scene):
    def construct(self):
        r = 2.4
        on = lambda d: r * np.array([np.cos(d * DEGREES), np.sin(d * DEGREES), 0])
        A, B, D = on(200), on(300), on(110)
        circ = Circle(radius=r, color=GREY).shift(LEFT * 1.5)
        A, B, D = A + LEFT * 1.5, B + LEFT * 1.5, D + LEFT * 1.5
        self.play(Create(circ))
        t = ValueTracker(20)
        C = lambda: on(t.get_value()) + LEFT * 1.5
        def quad():
            c = C(); E = c + 1.6 * (c - B) / np.linalg.norm(c - B)      # BC의 연장선 위의 점 E
            return VGroup(Polygon(A, B, c, D, color=WHITE), DashedLine(c, E, color=GREY),
                          Angle(Line(A, B), Line(A, D), radius=0.5, color=PINK),
                          Angle(Line(c, D), Line(c, B), radius=0.45, color=BLUE),
                          Angle(Line(c, E), Line(c, D), radius=0.7, color=PINK),
                          Text("A", font_size=24).next_to(A, LEFT), Text("B", font_size=24).next_to(B, DOWN),
                          Text("C", font_size=24).next_to(c, RIGHT), Text("D", font_size=24).next_to(D, UP),
                          Text("E", font_size=24).next_to(E, RIGHT))
        def panel():
            c = C(); E = c + (c - B)
            a, cc, ext = angle_deg(A, B, D), angle_deg(c, B, D), angle_deg(c, D, E)
            return Text(f"∠A = {a:3.0f}°\n∠C = {cc:3.0f}°\n∠A + ∠C = {a+cc:3.0f}°\n외각 ∠DCE = {ext:3.0f}°",
                        font="NanumGothic", font_size=28, line_spacing=1.2).move_to(RIGHT * 4)
        self.add(always_redraw(quad), always_redraw(panel))
        self.play(t.animate.set_value(80), run_time=4)
        self.play(t.animate.set_value(-40), run_time=4)
        self.wait(1.5)
```

## 🏆 도전 과제 — 크리스탈 앵글 💎

1. `CyclicQuad`에서 네 점의 각도 `200, 300, 20, 110`을 바꿔 다른 사각형을 만들고, 대각의 합이 여전히 180°인지 확인하세요.
2. 사각형에 보석처럼 색을 채우고(`set_fill`), 대각 두 개가 같은 색으로 빛나게 해 보세요.
3. (보너스) 네 점 중 하나를 원 **안쪽**(예: `0.6 * on(20)`)으로 옮기면 합이 180°가 아니게 되는 걸 확인해 보세요 — 내접해야만 성립하는 성질이에요!

```python
%%manim -qm -v WARNING MyCrystal

def angle_deg(V, A, B):
    u, w = A - V, B - V
    return np.degrees(np.arccos(np.clip(np.dot(u, w) / (np.linalg.norm(u) * np.linalg.norm(w)), -1, 1)))

class MyCrystal(Scene):
    def construct(self):
        r = 2.6
        on = lambda d: r * np.array([np.cos(d * DEGREES), np.sin(d * DEGREES), 0])
        A, B, C, D = on(200), on(300), on(20), on(110)      # 1. 각도를 바꿔 보세요 / 3. C = 0.6 * on(20)
        quad = Polygon(A, B, C, D, color=WHITE)              # 2. .set_fill(TEAL, opacity=0.4)
        self.play(Create(Circle(radius=r, color=GREY)), Create(quad))
        a, c = angle_deg(A, D, B), angle_deg(C, B, D)
        self.play(Write(Text(f"∠A + ∠C = {a:.0f}° + {c:.0f}° = {a+c:.0f}°", font_size=32, color=YELLOW).to_edge(UP)))
        self.wait(2)
```

## 📝 오늘 배운 것 정리

- 원에 내접하는 사각형: **∠A + ∠C = 180°, ∠B + ∠D = 180°** (두 호의 중심각 합 360°의 절반).
- **한 외각의 크기는 이웃한 내각의 대각의 크기와 같다**.
- 꼭짓점이 원 위에 있어야만 성립한다 — 원 안으로 들어가면 깨진다.
- `Sector`로 중심각 색칠, `DashedLine`으로 연장선, `always_redraw`로 움직이는 사각형.
