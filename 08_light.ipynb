## 1. 같은 호를 보는 원주각은 모두 같다

호 AB를 원 위의 여러 점 P₁, P₂, P₃에서 바라보면, 각 ∠AP₁B, ∠AP₂B, ∠AP₃B는 **전부 같은 크기**예요.
(모두 같은 중심각의 절반이니까!) 세 조명이 같은 호를 비추는 모습으로 그려 봐요.

```python
%%manim -qm -v WARNING SameArcSameAngle

def angle_deg(V, A, B):
    u, w = A - V, B - V
    return np.degrees(np.arccos(np.clip(np.dot(u, w) / (np.linalg.norm(u) * np.linalg.norm(w)), -1, 1)))

class SameArcSameAngle(Scene):
    def construct(self):
        r = 2.6
        on = lambda d: r * np.array([np.cos(d * DEGREES), np.sin(d * DEGREES), 0])
        A, B = on(-60), on(0)
        circ = Circle(radius=r, color=GREY)
        arc = Arc(radius=r, start_angle=-60 * DEGREES, angle=60 * DEGREES, color=YELLOW, stroke_width=10)
        self.play(Create(circ), Create(arc), FadeIn(Dot(A)), FadeIn(Dot(B)),
                  Write(Text("A", font_size=28).next_to(A, DR)), Write(Text("B", font_size=28).next_to(B, RIGHT)))
        colors = [BLUE, GREEN, PINK]
        for i, (d, color) in enumerate(zip([80, 140, 200], colors)):
            P = on(d)
            beam = Polygon(P, A, B, color=color, fill_opacity=0.25, stroke_width=0)     # 조명 빔
            ang = Angle(Line(P, A), Line(P, B), radius=0.5, color=color)
            label = Text(f"P{i+1}: {angle_deg(P, A, B):.0f}°", font_size=28, color=color).next_to(P, P / np.linalg.norm(P), buff=0.2)
            self.play(FadeIn(beam), FadeIn(Dot(P, color=color)), Create(ang), Write(label), run_time=1)
        msg = Text("같은 호 → 같은 원주각", font="NanumGothic", font_size=34, color=YELLOW).to_edge(UP)
        self.play(Write(msg))
        self.wait(2)
```

## 2. ⭐ 말없는 증명 — 호가 2배면 원주각도 2배

호 AB와 호 BC의 길이가 같으면 원주각도 같고, 호 AC(= 2배)에 대한 원주각은 **2배**예요.
점 P에서 세 호를 비추는 조명 각을 비교해 봅시다.

```python
%%manim -qm -v WARNING DoubleArcDoubleAngle

def angle_deg(V, A, B):
    u, w = A - V, B - V
    return np.degrees(np.arccos(np.clip(np.dot(u, w) / (np.linalg.norm(u) * np.linalg.norm(w)), -1, 1)))

class DoubleArcDoubleAngle(Scene):
    def construct(self):
        r = 2.6
        on = lambda d: r * np.array([np.cos(d * DEGREES), np.sin(d * DEGREES), 0])
        A, B, C, P = on(-70), on(-20), on(30), on(160)
        circ = Circle(radius=r, color=GREY)
        self.play(Create(circ), *[FadeIn(Dot(q)) for q in (A, B, C, P)],
                  Write(Text("A", font_size=26).next_to(A, DR)), Write(Text("B", font_size=26).next_to(B, RIGHT)),
                  Write(Text("C", font_size=26).next_to(C, UR)), Write(Text("P", font_size=26).next_to(P, LEFT)))
        arc1 = Arc(radius=r, start_angle=-70 * DEGREES, angle=50 * DEGREES, color=YELLOW, stroke_width=10)
        arc2 = Arc(radius=r, start_angle=-20 * DEGREES, angle=50 * DEGREES, color=GREEN, stroke_width=10)
        self.play(Create(arc1), Create(arc2))

        a1 = Angle(Line(P, A), Line(P, B), radius=0.7, color=YELLOW)
        a2 = Angle(Line(P, B), Line(P, C), radius=0.9, color=GREEN)
        rays = VGroup(Line(P, A, color=YELLOW), Line(P, B, color=WHITE), Line(P, C, color=GREEN))
        self.play(Create(rays), Create(a1), Create(a2))
        t1 = Text(f"호 AB → {angle_deg(P, A, B):.0f}°", font="NanumGothic", font_size=28, color=YELLOW).to_corner(UR)
        t2 = Text(f"호 BC → {angle_deg(P, B, C):.0f}°", font="NanumGothic", font_size=28, color=GREEN).next_to(t1, DOWN, aligned_edge=RIGHT)
        t3 = Text(f"호 AC → {angle_deg(P, A, C):.0f}°  (2배!)", font="NanumGothic", font_size=28, color=WHITE).next_to(t2, DOWN, aligned_edge=RIGHT)
        self.play(Write(t1), Write(t2)); self.wait(0.5); self.play(Write(t3))
        msg = Text("원주각은 호의 길이에 정비례", font="NanumGothic", font_size=34, color=YELLOW).to_edge(DOWN)
        self.play(Write(msg))
        self.wait(2)
```

## 3. 조명 각을 돌리면 호의 길이는? 💡

조명(원주각)의 크기를 `ValueTracker`로 키우면, 비추는 호의 길이가 **비례해서** 길어져요.
원주각 θ가 비추는 호의 중심각은 2θ, 호의 길이는 2πr × 2θ/360.

```python
%%manim -qm -v WARNING SpotlightArc

class SpotlightArc(Scene):
    def construct(self):
        r = 2.5
        P = np.array([0, -r, 0])                         # 조명 위치 (원 맨 아래)
        circ = Circle(radius=r, color=GREY)
        self.play(Create(circ), FadeIn(Dot(P, color=WHITE)), Write(Text("조명 P", font="NanumGothic", font_size=26).next_to(P, DOWN)))
        theta = ValueTracker(10)                          # 원주각(도)
        def arc():
            th = theta.get_value() * DEGREES
            return Arc(radius=r, start_angle=PI / 2 - th, angle=2 * th, color=YELLOW, stroke_width=10)   # 중심각 2θ
        def beam():
            th = theta.get_value() * DEGREES
            L = r * np.array([np.cos(PI / 2 + th), np.sin(PI / 2 + th), 0]); R_ = r * np.array([np.cos(PI / 2 - th), np.sin(PI / 2 - th), 0])
            return Polygon(P, L, R_, color=YELLOW, fill_opacity=0.25, stroke_width=0)
        def panel():
            th = theta.get_value()
            return Text(f"원주각 {th:4.0f}°   호의 길이 {2*PI*r*2*th/360:4.2f}", font="NanumGothic", font_size=30).to_edge(UP)
        self.add(always_redraw(beam), always_redraw(arc), always_redraw(panel))
        self.play(theta.animate.set_value(30), run_time=2); self.wait(0.4)
        self.play(theta.animate.set_value(60), run_time=2)   # 2배 → 호도 2배
        self.wait(2)
```

## 🏆 도전 과제 — 강아지와 고양이를 함께 비추기 🐶🐱

빛의 도시 게임처럼, 원 위에 **강아지**(점 D)와 **고양이**(점 C)가 있어요. 조명 P의 각을 조절해 둘을 동시에 비추세요.

1. `dog, cat`의 위치(각도)를 바꿔 보고, 둘을 모두 비추는 최소 원주각 `theta`가 몇 도인지 계산해 넣으세요. (힌트: 두 점 사이 중심각의 절반)
2. 조명 색과 빔의 투명도를 바꿔 보세요.
3. (보너스) 조명 P의 위치를 원 위 다른 곳으로 옮겨도 필요한 각이 같은지 확인해 보세요.

```python
%%manim -qm -v WARNING DogAndCat

class DogAndCat(Scene):
    def construct(self):
        r = 2.5
        on = lambda d: r * np.array([np.cos(d * DEGREES), np.sin(d * DEGREES), 0])
        dog, cat = 60, 120                 # 1. 위치를 바꿔 보세요
        theta = 30                         # 1. 둘 다 비추는 원주각 (도)
        P = on(-90)                        # 3. 조명 위치
        self.play(Create(Circle(radius=r, color=GREY)),
                  Write(Text("🐶", font_size=40).move_to(on(dog) * 1.15)), Write(Text("🐱", font_size=40).move_to(on(cat) * 1.15)),
                  FadeIn(Dot(P, color=WHITE)))
        mid = (dog + cat) / 2
        L, R_ = on(mid + theta), on(mid - theta)           # 조명이 비추는 호의 양 끝 (중심각 2θ)
        beam = Polygon(P, L, R_, color=YELLOW, fill_opacity=0.3, stroke_width=0)   # 2. 색·투명도
        arc = Arc(radius=r, start_angle=(mid - theta) * DEGREES, angle=2 * theta * DEGREES, color=YELLOW, stroke_width=10)
        self.play(FadeIn(beam), Create(arc))
        ok = theta * 2 >= abs(cat - dog)
        msg = Text("둘 다 밝아졌어요! ✨" if ok else "각이 부족해요…", font="NanumGothic", font_size=34,
                   color=GREEN if ok else ORANGE).to_edge(UP)
        self.play(Write(msg))
        self.wait(2)
```

## 📝 오늘 배운 것 정리

- 한 원에서 **같은 호(길이가 같은 호)에 대한 원주각은 모두 같다**.
- 원주각은 **호의 길이에 정비례**한다 (호 2배 → 원주각 2배). ※ 현의 길이와는 정비례하지 않아요!
- 원주각 θ의 조명은 중심각 2θ만큼의 호를 비춘다.
- `Polygon`으로 빛 빔, `Arc` + `ValueTracker`로 자라는 호를 만들었다.
