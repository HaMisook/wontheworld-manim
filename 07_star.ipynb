## 1. 중심각과 원주각

호 AB에 대해
- **중심각** ∠AOB: 꼭짓점이 중심 O
- **원주각** ∠APB: 꼭짓점 P가 원 위

각의 크기는 벡터의 내적으로 계산할 수 있어요. 아래 `angle_deg(P, A, B)`는 ∠APB를 도(°) 단위로 돌려줘요.

```python
%%manim -qm -v WARNING CentralInscribed

def angle_deg(V, A, B):
    u, w = A - V, B - V
    return np.degrees(np.arccos(np.dot(u, w) / (np.linalg.norm(u) * np.linalg.norm(w))))

class CentralInscribed(Scene):
    def construct(self):
        r = 2.6
        O = ORIGIN
        A = r * np.array([np.cos(-30 * DEGREES), np.sin(-30 * DEGREES), 0])
        B = r * np.array([np.cos(70 * DEGREES), np.sin(70 * DEGREES), 0])
        P = r * np.array([np.cos(190 * DEGREES), np.sin(190 * DEGREES), 0])
        circ = Circle(radius=r, color=GREY)
        arc = Arc(radius=r, start_angle=-30 * DEGREES, angle=100 * DEGREES, color=YELLOW, stroke_width=8)
        self.play(Create(circ), Create(arc), FadeIn(Dot(O, color=ORANGE)),
                  *[FadeIn(Dot(q)) for q in (A, B, P)],
                  Write(Text("A", font_size=28).next_to(A, DR)), Write(Text("B", font_size=28).next_to(B, UR)),
                  Write(Text("P", font_size=28).next_to(P, LEFT)), Write(Text("O", font_size=28).next_to(O, DOWN)))

        OA, OB = Line(O, A, color=ORANGE), Line(O, B, color=ORANGE)
        central = Angle(OA, OB, radius=0.6, color=ORANGE)
        tc = Text(f"중심각 {angle_deg(O, A, B):.0f}°", font="NanumGothic", font_size=28, color=ORANGE).to_corner(UR)
        self.play(Create(OA), Create(OB), Create(central), Write(tc))

        PA, PB = Line(P, A, color=BLUE), Line(P, B, color=BLUE)
        insc = Angle(PA, PB, radius=0.7, color=BLUE)
        ti = Text(f"원주각 {angle_deg(P, A, B):.0f}°", font="NanumGothic", font_size=28, color=BLUE).next_to(tc, DOWN, aligned_edge=RIGHT)
        self.play(Create(PA), Create(PB), Create(insc), Write(ti))
        msg = Text("원주각 = 중심각 ÷ 2", font="NanumGothic", font_size=34, color=YELLOW).to_edge(DOWN)
        self.play(Write(msg))
        self.wait(2)
```

## 2. ⭐ 말없는 증명 — P가 움직여도 원주각은 그대로

점 P를 호 AB 반대편 원 위에서 움직여 봐요. 원주각 ∠APB는 **한 번도 변하지 않고**, 늘 중심각의 절반이에요.
숫자판이 두 각을 동시에 보여 줍니다.

```python
%%manim -qm -v WARNING MovingInscribed

def angle_deg(V, A, B):
    u, w = A - V, B - V
    return np.degrees(np.arccos(np.clip(np.dot(u, w) / (np.linalg.norm(u) * np.linalg.norm(w)), -1, 1)))

class MovingInscribed(Scene):
    def construct(self):
        r = 2.4
        O = ORIGIN
        a1, a2 = -40, 60                                        # 호 AB
        A = r * np.array([np.cos(a1 * DEGREES), np.sin(a1 * DEGREES), 0])
        B = r * np.array([np.cos(a2 * DEGREES), np.sin(a2 * DEGREES), 0])
        circ = Circle(radius=r, color=GREY)
        arc = Arc(radius=r, start_angle=a1 * DEGREES, angle=(a2 - a1) * DEGREES, color=YELLOW, stroke_width=8)
        self.play(Create(circ), Create(arc), FadeIn(Dot(O, color=ORANGE)), FadeIn(Dot(A)), FadeIn(Dot(B)))
        self.play(Create(Line(O, A, color=ORANGE)), Create(Line(O, B, color=ORANGE)),
                  Create(Angle(Line(O, A), Line(O, B), radius=0.5, color=ORANGE)))

        t = ValueTracker(120)
        def P():
            a = t.get_value() * DEGREES
            return r * np.array([np.cos(a), np.sin(a), 0])
        def rays():
            p = P()
            return VGroup(Line(p, A, color=BLUE), Line(p, B, color=BLUE), Dot(p, color=BLUE),
                          Angle(Line(p, A), Line(p, B), radius=0.5, color=BLUE, other_angle=False))
        def panel():
            return Text(f"중심각 {angle_deg(O, A, B):.0f}°\n원주각 {angle_deg(P(), A, B):.0f}°",
                        font="NanumGothic", font_size=30, line_spacing=1.2).to_corner(UR)
        self.add(always_redraw(rays), always_redraw(panel))
        self.play(t.animate.set_value(300), run_time=6, rate_func=linear)
        self.play(t.animate.set_value(120), run_time=4)
        msg = Text("P가 어디 있든 원주각은 중심각의 절반", font="NanumGothic", font_size=30, color=YELLOW).to_edge(DOWN)
        self.play(Write(msg))
        self.wait(2)
```

## 3. 반원에 대한 원주각은 90°

호 AB가 **반원**이면 중심각이 180°이므로 원주각은 90°!
지름 AB 위에 어떤 점 P를 잡아도 ∠APB는 직각이에요. (얼음의 도시에서 본 '빗변 = 지름'과 같은 이야기)

```python
%%manim -qm -v WARNING SemicircleAngle

class SemicircleAngle(Scene):
    def construct(self):
        r = 2.5
        A, B = np.array([-r, 0, 0]), np.array([r, 0, 0])
        circ = Circle(radius=r, color=GREY)
        self.play(Create(circ), Create(Line(A, B, color=YELLOW)), FadeIn(Dot(ORIGIN, color=ORANGE)),
                  Write(Text("A", font_size=28).next_to(A, LEFT)), Write(Text("B", font_size=28).next_to(B, RIGHT)))
        t = ValueTracker(30)
        def P():
            a = t.get_value() * DEGREES
            return r * np.array([np.cos(a), np.sin(a), 0])
        pic = always_redraw(lambda: VGroup(Line(P(), A, color=BLUE), Line(P(), B, color=BLUE), Dot(P(), color=BLUE),
                                            RightAngle(Line(P(), A), Line(P(), B), length=0.35, color=RED)))
        self.add(pic)
        self.play(t.animate.set_value(160), run_time=4)
        self.play(t.animate.set_value(90), run_time=2)
        msg = Text("중심각 180° → 원주각 90°", font="NanumGothic", font_size=34, color=YELLOW).to_edge(DOWN)
        self.play(Write(msg))
        self.wait(2)
```

## 🏆 도전 과제 — 별을 조준하라 ⭐

`MovingInscribed`를 바탕으로 한 아래 셀에서

1. 호 AB의 크기 `a1, a2`를 바꿔 중심각을 120°로 만들어 보세요. 원주각은 몇 도가 되나요?
2. 원주각 꼭짓점 P를 노란 별(`Star(color=YELLOW)`)로 바꿔 보세요.
3. (보너스) P가 호 AB **쪽**(짧은 호 위)으로 가면 원주각은 어떻게 되나요? `t` 범위를 바꿔 실험해 보세요.

```python
%%manim -qm -v WARNING StarSniper

def angle_deg(V, A, B):
    u, w = A - V, B - V
    return np.degrees(np.arccos(np.clip(np.dot(u, w) / (np.linalg.norm(u) * np.linalg.norm(w)), -1, 1)))

class StarSniper(Scene):
    def construct(self):
        r = 2.4
        a1, a2 = -40, 60                    # 1. 중심각이 120°가 되도록 바꿔 보세요
        A = r * np.array([np.cos(a1 * DEGREES), np.sin(a1 * DEGREES), 0])
        B = r * np.array([np.cos(a2 * DEGREES), np.sin(a2 * DEGREES), 0])
        self.play(Create(Circle(radius=r, color=GREY)), FadeIn(Dot(A)), FadeIn(Dot(B)),
                  Create(Arc(radius=r, start_angle=a1 * DEGREES, angle=(a2 - a1) * DEGREES, color=YELLOW, stroke_width=8)))
        t = ValueTracker(120)
        def P():
            a = t.get_value() * DEGREES
            return r * np.array([np.cos(a), np.sin(a), 0])
        marker = always_redraw(lambda: Dot(P(), color=BLUE))          # 2. Star(color=YELLOW).scale(0.3).move_to(P()) 로!
        rays = always_redraw(lambda: VGroup(Line(P(), A, color=BLUE), Line(P(), B, color=BLUE)))
        panel = always_redraw(lambda: Text(f"원주각 {angle_deg(P(), A, B):.0f}°", font="NanumGothic", font_size=32).to_corner(UR))
        self.add(rays, marker, panel)
        self.play(t.animate.set_value(300), run_time=6, rate_func=linear)   # 3. 범위를 바꿔 보세요
        self.wait(1)
```

## 📝 오늘 배운 것 정리

- 한 호에 대한 **원주각 = 중심각의 ½**, 그래서 같은 호에 대한 원주각은 P가 어디 있든 같다.
- 반원(지름)에 대한 원주각은 **90°**.
- 각의 크기는 내적으로: `arccos(u·w / |u||w|)`. `Angle`, `RightAngle`, `Star`를 썼다.
