## 1. 접선과 현이 이루는 각 = 원주각

점 A에서의 접선과 현 AB가 이루는 각 ∠BAT는, 그 각 안쪽 호 AB에 대한 원주각 ∠APB와 **같아요**.
(접선은 반지름 OA에 수직이니, ∠BAT = 90° − ∠OAB = ½∠AOB = 원주각)

````python
%%manim -qm -v WARNING TangentChordAngle

def angle_deg(V, A, B):
    u, w = A - V, B - V
    return np.degrees(np.arccos(np.clip(np.dot(u, w) / (np.linalg.norm(u) * np.linalg.norm(w)), -1, 1)))

class TangentChordAngle(Scene):
    def construct(self):
        r = 2.4
        on = lambda d: r * np.array([np.cos(d * DEGREES), np.sin(d * DEGREES), 0])
        A, B, P = on(-90), on(10), on(150)
        T = A + 3 * np.array([1, 0, 0])                     # 접선 위의 점 T (A에서 오른쪽)
        circ = Circle(radius=r, color=GREY)
        tangent = Line(A - 3.5 * RIGHT, A + 3.5 * RIGHT, color=BLUE)
        self.play(Create(circ), Create(tangent), FadeIn(Dot(A, color=BLUE)), FadeIn(Dot(B)), FadeIn(Dot(P)),
                  Write(Text("A", font_size=28).next_to(A, DOWN)), Write(Text("B", font_size=28).next_to(B, RIGHT)),
                  Write(Text("P", font_size=28).next_to(P, UL)), Write(Text("T", font_size=28, color=BLUE).next_to(T, DOWN)))
        chord = Line(A, B, color=YELLOW)
        arc = Arc(radius=r, start_angle=-90 * DEGREES, angle=100 * DEGREES, color=YELLOW, stroke_width=8)
        self.play(Create(chord), Create(arc))

        a1 = Angle(Line(A, T), Line(A, B), radius=0.6, color=BLUE)
        t1 = Text(f"접선과 현이 이루는 각 {angle_deg(A, T, B):.0f}°", font="NanumGothic", font_size=26, color=BLUE).to_corner(UL)
        self.play(Create(a1), Write(t1))
        a2 = Angle(Line(P, B), Line(P, A), radius=0.5, color=PINK)
        t2 = Text(f"호 AB의 원주각 {angle_deg(P, A, B):.0f}°", font="NanumGothic", font_size=26, color=PINK).next_to(t1, DOWN, aligned_edge=LEFT)
        self.play(Create(Line(P, A, color=PINK)), Create(Line(P, B, color=PINK)), Create(a2), Write(t2))
        msg = Text("두 각은 같다 — 봉인 해제!", font="NanumGothic", font_size=34, color=YELLOW).to_edge(DOWN)
        self.play(Write(msg))
        self.wait(2)
````

## 2. ⭐ 말없는 증명 — 현을 돌리면 두 각이 함께 변한다

점 B를 원 위에서 움직이면 접선과 현이 이루는 각도, 원주각도 같이 변하지만 **늘 같은 값**이에요.
B가 A의 정반대(지름)로 가면 두 각은 **90°**가 돼요.

````python
%%manim -qm -v WARNING RotatingChord

def angle_deg(V, A, B):
    u, w = A - V, B - V
    return np.degrees(np.arccos(np.clip(np.dot(u, w) / (np.linalg.norm(u) * np.linalg.norm(w)), -1, 1)))

class RotatingChord(Scene):
    def construct(self):
        r = 2.3
        on = lambda d: r * np.array([np.cos(d * DEGREES), np.sin(d * DEGREES), 0])
        A, P = on(-90), on(170)
        T = A + 3 * RIGHT
        self.play(Create(Circle(radius=r, color=GREY)), Create(Line(A - 3.5 * RIGHT, A + 3.5 * RIGHT, color=BLUE)),
                  FadeIn(Dot(A, color=BLUE)), FadeIn(Dot(P)), Write(Text("A", font_size=26).next_to(A, DOWN)), Write(Text("P", font_size=26).next_to(P, LEFT)))
        t = ValueTracker(-20)
        B = lambda: on(t.get_value())
        def pic():
            b = B()
            return VGroup(Line(A, b, color=YELLOW, stroke_width=6), Dot(b, color=YELLOW),
                          Angle(Line(A, T), Line(A, b), radius=0.6, color=BLUE),
                          Line(P, A, color=PINK), Line(P, b, color=PINK), Angle(Line(P, A), Line(P, b), radius=0.5, color=PINK))
        def panel():
            b = B()
            return Text(f"접선·현 각 {angle_deg(A, T, b):3.0f}°\n원주각    {angle_deg(P, A, b):3.0f}°", font="NanumGothic",
                        font_size=30, line_spacing=1.2).to_corner(UR)
        self.add(always_redraw(pic), always_redraw(panel))
        self.play(t.animate.set_value(90), run_time=5)     # B가 A의 정반대(지름)까지
        msg = Text("90°가 되면 현 AB는 지름!", font="NanumGothic", font_size=32, color=YELLOW).to_edge(DOWN)
        self.play(Write(msg))
        self.wait(2)
````
