## 1. 중심에서 현에 내린 수선

현 AB에 원의 중심 O에서 수선을 내려 발을 M이라 하면 **AM = MB**예요.
(△OAM과 △OBM이 합동이기 때문 — OA = OB는 반지름, OM은 공통, ∠M = 90°)

아래 코드는 AM과 MB의 길이를 재서 같음을 보여 줘요.

```python
%%manim -qm -v WARNING PerpFromCenter

class PerpFromCenter(Scene):
    def construct(self):
        r = 2.6
        circ = Circle(radius=r, color=GREY)
        O = ORIGIN
        a1, a2 = 200 * DEGREES, 340 * DEGREES
        A = r * np.array([np.cos(a1), np.sin(a1), 0])
        B = r * np.array([np.cos(a2), np.sin(a2), 0])
        M = (A + B) / 2
        self.play(Create(circ), FadeIn(Dot(O, color=ORANGE)), Write(Text("O", font_size=28).next_to(O, UP)))

        chord = Line(A, B, color=YELLOW)
        self.play(Create(chord), FadeIn(Dot(A)), FadeIn(Dot(B)),
                  Write(Text("A", font_size=28).next_to(A, DL)), Write(Text("B", font_size=28).next_to(B, DR)))

        perp = Line(O, M, color=BLUE)
        ra = RightAngle(perp, chord, length=0.3, quadrant=(-1, 1), color=BLUE)
        self.play(Create(perp), Create(ra), FadeIn(Dot(M, color=BLUE)), Write(Text("M", font_size=28, color=BLUE).next_to(M, DOWN)))

        # 두 반지름 OA, OB → 합동 삼각형
        t1 = Polygon(O, A, M, color=GREEN, fill_opacity=0.3)
        t2 = Polygon(O, B, M, color=PINK, fill_opacity=0.3)
        self.play(FadeIn(t1), FadeIn(t2))
        self.play(t1.copy().animate.become(t2), run_time=1.5)     # 뒤집어 포개기

        b1 = BraceBetweenPoints(A, M, direction=DOWN, color=GREEN)
        b2 = BraceBetweenPoints(M, B, direction=DOWN, color=PINK)
        self.play(GrowFromCenter(b1), GrowFromCenter(b2))
        msg = Text("AM = MB — 중심에서 내린 수선은 현을 이등분!", font="NanumGothic", font_size=30).to_edge(UP)
        self.play(Write(msg))
        self.wait(2)
```

## 2. ⭐ 말없는 증명 — 현의 수직이등분선은 중심을 지난다

거꾸로 생각해 봐요. 아무 현이나 그어서 **수직이등분선**을 그으면, 그 선은 반드시 원의 중심을 지나요.
현을 3개 그어 세 수직이등분선이 **모두 중심 한 점**을 지나는 것을 확인해요.

```python
%%manim -qm -v WARNING BisectorsThroughCenter

def perp_bisector(P, Q, length=7):
    mid = (P + Q) / 2
    d = Q - P
    n = np.array([-d[1], d[0], 0]) / np.linalg.norm(d)
    return Line(mid - n * length / 2, mid + n * length / 2)

class BisectorsThroughCenter(Scene):
    def construct(self):
        r = 2.6
        circ = Circle(radius=r, color=GREY)
        self.play(Create(circ))
        chords = [(10, 120, YELLOW), (150, 250, GREEN), (280, 20, PINK)]
        for a, b, color in chords:
            A = r * np.array([np.cos(a * DEGREES), np.sin(a * DEGREES), 0])
            B = r * np.array([np.cos(b * DEGREES), np.sin(b * DEGREES), 0])
            chord = Line(A, B, color=color)
            bis = DashedVMobject(perp_bisector(A, B).set_color(color))
            self.play(Create(chord), run_time=0.6)
            self.play(Create(bis), run_time=0.8)
        O = Dot(ORIGIN, color=ORANGE).scale(1.5)
        msg = Text("세 수직이등분선 모두 중심 O를 지난다", font="NanumGothic", font_size=30, color=ORANGE).to_edge(UP)
        self.play(FadeIn(O), Write(msg))
        self.wait(2)
```

## 3. 깨진 접시 복원하기 🍽️

접시 조각(호의 일부)만 남았어요. 원래 접시(원)를 되살릴 수 있을까요?

1. 조각 위에 점 3개 A, B, C를 잡는다.
2. 현 AB와 현 BC의 **수직이등분선**을 긋는다.
3. 두 선이 만나는 점이 **원래 접시의 중심**! 중심에서 A까지의 거리가 반지름.

```python
%%manim -qm -v WARNING BrokenPlate

def perp_bisector(P, Q, length=9):
    mid = (P + Q) / 2
    d = Q - P
    n = np.array([-d[1], d[0], 0]) / np.linalg.norm(d)
    return Line(mid - n * length / 2, mid + n * length / 2)

class BrokenPlate(Scene):
    def construct(self):
        R = 2.8
        center = np.array([0, -1.2, 0])                   # 숨겨진 원래 중심
        piece = Arc(radius=R, start_angle=50 * DEGREES, angle=80 * DEGREES,
                    color=WHITE, stroke_width=10).shift(center)
        title = Text("깨진 접시 조각… 원래 모양은?", font="NanumGothic", font_size=32).to_edge(UP)
        self.play(Write(title), Create(piece))

        pts = [center + R * np.array([np.cos(d * DEGREES), np.sin(d * DEGREES), 0]) for d in (55, 90, 125)]
        A, B, C = pts
        self.play(*[FadeIn(Dot(p, color=YELLOW)) for p in pts],
                  Write(Text("A", font_size=26).next_to(A, RIGHT)), Write(Text("B", font_size=26).next_to(B, UP)),
                  Write(Text("C", font_size=26).next_to(C, LEFT)))
        c1, c2 = Line(A, B, color=YELLOW), Line(B, C, color=YELLOW)
        self.play(Create(c1), Create(c2))

        l1, l2 = perp_bisector(A, B).set_color(BLUE), perp_bisector(B, C).set_color(GREEN)
        b1, b2 = DashedVMobject(l1), DashedVMobject(l2)
        self.play(Create(b1)); self.play(Create(b2))

        O = line_intersection(l1.get_start_and_end(), l2.get_start_and_end())
        dotO = Dot(O, color=ORANGE).scale(1.4)
        self.play(FadeIn(dotO), Write(Text("중심!", font="NanumGothic", font_size=28, color=ORANGE).next_to(dotO, DOWN)))
        plate = Circle(radius=np.linalg.norm(A - O), color=WHITE, stroke_width=4).move_to(O)
        self.play(Create(plate), run_time=2)
        self.play(FadeOut(b1), FadeOut(b2), FadeOut(c1), FadeOut(c2))
        self.play(Transform(title, Text("접시 복원 완료! 🍽️", font="NanumGothic", font_size=36, color=GREEN).to_edge(UP)))
        self.wait(2)
```

## 🏆 도전 과제 — 내 접시 조각으로

1. `BrokenPlate`에서 조각의 크기 `angle=80 * DEGREES`를 `40 * DEGREES`로 줄여도 복원되나요? (점 3개의 각도도 조각 안에 있도록 `(55, 90, 125)`를 `(55, 70, 85)`처럼 바꾸세요.)
2. 숨겨진 중심 `center`와 반지름 `R`을 바꿔서 다른 접시를 만들어 보세요.
3. (보너스) 복원된 접시에 `set_fill`로 색을 칠하고 무늬(작은 원)를 넣어 보세요.

```python
%%manim -qm -v WARNING MyPlate

def perp_bisector(P, Q, length=9):
    mid = (P + Q) / 2
    d = Q - P
    n = np.array([-d[1], d[0], 0]) / np.linalg.norm(d)
    return Line(mid - n * length / 2, mid + n * length / 2)

class MyPlate(Scene):
    def construct(self):
        R = 2.8                                  # 2. 반지름을 바꿔 보세요
        center = np.array([0, -1.2, 0])          # 2. 중심을 바꿔 보세요
        piece_angle = 80                         # 1. 조각 크기
        angles = (55, 90, 125)                   # 1. 점 3개의 위치(조각 안에!)

        piece = Arc(radius=R, start_angle=50 * DEGREES, angle=piece_angle * DEGREES,
                    color=WHITE, stroke_width=10).shift(center)
        self.play(Create(piece))
        A, B, C = [center + R * np.array([np.cos(d * DEGREES), np.sin(d * DEGREES), 0]) for d in angles]
        l1, l2 = perp_bisector(A, B).set_color(BLUE), perp_bisector(B, C).set_color(GREEN)
        self.play(Create(DashedVMobject(l1)), Create(DashedVMobject(l2)))
        O = line_intersection(l1.get_start_and_end(), l2.get_start_and_end())
        plate = Circle(radius=np.linalg.norm(A - O), color=WHITE).move_to(O)
        self.play(FadeIn(Dot(O, color=ORANGE)), Create(plate))
        # 3. plate.set_fill(색, opacity=0.3) 을 넣어 보세요
        self.wait(2)
```

## 📝 오늘 배운 것 정리

- 중심에서 현에 내린 수선은 현을 **수직이등분**한다 (AM = MB).
- 현의 **수직이등분선은 항상 원의 중심을 지난다** → 현 2개면 중심을 찾을 수 있다.
- 깨진 접시(호 조각)도 점 3개 → 수직이등분선 2개 → 중심 → 원 복원!
- `RightAngle`, `BraceBetweenPoints`, `line_intersection`, `Transform`을 썼다.
